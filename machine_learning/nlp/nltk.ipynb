### **Removing Stopwords**

Stopwords are words that you want to ignore, so you filter them out of your text when you’re processing it. Very common words like 'in', 'is', and 'an' are often used as stop words since they don’t add a lot of meaning to a text in and of themselves.

```python
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))

print("Number of stop words:", len(stop_words))
print(list(stop_words)[:10])
```

```python
stop_words.update(['@',"'",'.','"','/','!',',',"'ve","...","n't",'$',"'s"])
print("Number of stop words after update:", len(stop_words))
```

```python
data = []
for line in df.loc[:2,'text'].str.lower():
    processed_line = []
    for word in nltk.word_tokenize(line):
        if word not in stop_words:
            processed_line.append(word)
    data.append(processed_line)
print(data)
```

### **Stemming**
Reduce words to their root, which is the core part of a word. Take note that the core part ('root') may not be a complete English word.

For example, the words “helping” and “helped” share the root “help.”

There are two stemmers available in NLTK, 
- **PorterStemmer()**  
- **SnowballStemmer()**. The Snowball stemmer, which is also called Porter2, is an improvement on the original Porter stemmer.

```python
from nltk.stem import PorterStemmer
p_stemmer = PorterStemmer()

print(p_stemmer.stem('reading'))
```

```python
list(map(p_stemmer.stem, ['help', 'helped', 'helping', 'helps', 'helper', 'helpful']))
```

```python
from nltk.stem import SnowballStemmer
s_stemmer = SnowballStemmer('english')

list(map(p_stemmer.stem, ['help', 'helped', 'helping', 'helps', 'helper', 'helpful']))
```

```python
from nltk import word_tokenize

words_in_each_string = []
for each_string in df.loc[:100,'text'].str.lower():
    words_in_each_string.append(word_tokenize(each_string))

original_words_list = []
p_stemmed_list = []
s_stemmed_list = []
for each_list in words_in_each_string:        
    for each_word in each_list:
        original_words_list.append(each_word)
        p_stemmed_list.append(p_stemmer.stem(each_word))
        s_stemmed_list.append(s_stemmer.stem(each_word))        
```

```python
data = pd.DataFrame({"original_words":original_words_list, "porter_stemmed":p_stemmed_list, "snowball_stemmed":s_stemmed_list})
data.shape
```

```python
data.drop_duplicates(inplace=True)
data.shape
```

### **Lemmatization**
Stemming is a process that stems or removes last few characters from a word, often leading to incorrect meanings and spelling. Lemmatization considers the context and converts the word to its meaningful base form, which is called Lemma.

For instance, stemming the word 'amazed' would return 'amaz'. However, lemmatizing the word 'amazed' would give 'amaze'

```python
print(p_stemmer.stem('amazed'))
print(s_stemmer.stem('amazed'))
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatizer.lemmatize('amazed')
```

## **N-Grams**

```python
from nltk.util import ngrams
for each_string in df.loc[:10,'text'].str.lower():
    tokens = each_string.split()
    bigrams = ngrams(tokens, 2)
    print(list(bigrams))
```

```python
import spacy
# Loading the model
nlp=spacy.load('en_core_web_sm')
```

```python
text = "Listen! There is a well-coronated launch today at 10:00 AM from the main hall."

# Creating Doc object
doc=nlp(text)

# Getting dependency tags
for token in doc:
    print(token.text,'=>',token.dep_)
```

```python
# Importing visualizer
from spacy import displacy
# Visualizing dependency tree
displacy.render(doc,jupyter=True)
```

## Chunking

```python
chunker = RegexpParser("""
                       NP: {<NN.*>+}  #To extract Noun Phrases
                       P: {<IN>}      #To extract Prepositions
                       V: {<VB.*>}    #To extract Verbs
                       PP: {<P> <NP>} #To extract Prepositional Phrases
                       VP: {<V> <NP>} #To extract Verb Phrases
""")

result = chunker.parse(tags)    

print('Our Parse Notation:n', result)
```
